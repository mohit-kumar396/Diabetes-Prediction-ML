```python
import pandas as pd
import sqlite3
customers_df=pd.read_csv("customers.csv")
orders_df=pd.read_csv("orders.csv")
conn=sqlite3.connect(":memory:")
```

```python
customers_df.to_sql("customers", conn,index=False, if_exists="replace")
orders_df.to_sql("orders", conn,index=False, if_exists="replace")
# Above line ka meaning:- "Take the Pandas DataFrame df and store it as an SQL
# table named orders inside the database conn."
```

```python
query1="""
select * from  orders
"""
result=pd.read_sql(query1,conn)
result
```

```python
query2="""
select product, amount from orders where customer_id=101 or customer_id=103
"""
result=pd.read_sql(query2,conn)
result
```

```python
query3="""
select * from  customers
"""
result=pd.read_sql(query3,conn)
result
```

```python
# 1. Inner Join
```

```python
query4="""
select customers.customer_name, customers.city, orders.product, orders.amount
from customers inner join orders on customers.customer_id=orders.customer_id
"""
result=pd.read_sql(query4,conn)
result
```

#**Left Join**

```python
query5="""
select customers.customer_name, customers.city, orders.product, orders.amount
from customers left join orders on customers.customer_id=orders.customer_id
"""
result=pd.read_sql(query5,conn)
result
```

```python
query6="""
select customers.customer_name, customers.city, orders.product, orders.amount
from customers left join orders on customers.customer_id=orders.customer_id
"""
result=pd.read_sql(query6,conn)
result
```

#**Right Join:-  returns all rows from the right table (table2), and only the matched rows from the left table (table1).**

```python
# iss query ko left join ko reverse kar kk likha hai bcz right join and full join sqlite me suppoert nhi karta.
query7="""
select customers.customer_name, customers.city, orders.product, orders.amount
from orders left join customers on customers.customer_id=orders.customer_id
"""
result=pd.read_sql(query7,conn)
result
```

#**Full join:- Since Full join is not supported in sqlite so we will do union of left join and right join(reversed left join)**

```python
query8="""
select customers.customer_name, customers.city, orders.product, orders.amount
from customers left join orders on customers.customer_id=orders.customer_id
union
select customers.customer_name, customers.city, orders.product, orders.amount
from orders left join customers on customers.customer_id=orders.customer_id
"""
result=pd.read_sql(query8,conn)
result
```
