Array Reshaping -> Reshape, Ravel, flatten

```python
# reshape() -> changes the shape of a NumPy array without changing its data.
# array.reshape(new_shape)
# array → existing NumPy array
arr = np.array([1, 2, 3, 4, 5, 6])

reshaped = arr.reshape(2, 3)
print(reshaped)
```

```python
reshape2 = reshaped.reshape((3,2))
```

```python
reshape2
```

```python
# ravel() -> converts a multi-dimensional array into a 1D array.
# syntax :- array.ravel()
# array → existing NumPy array
ravel = reshape2.ravel()

ravel
```

```python
# Note:
# ravel() usually returns a view of the original array instead of a separate copy.
# Therefore, changes made to the raveled array may also affect the original array..
ravel[0] = 100
print(ravel)
print(reshape2)
```

```python
# flatten() -> converts a multi-dimensional array into a 1D array.
# It always returns a copy, so changes to the flattened array
# do not affect the original array.
# Syntax :-
# array.flatten()
# array → existing NumPy array
flat = reshape2.flatten()
flat
```

```python
flat[0] = 1
```

```python
flat
```

```python
reshape2
# See Reshape change nhi hua hai
```
